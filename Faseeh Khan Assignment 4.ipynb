### Q1. Stop sending emails after 3 users (for + break)

```python
users = ['Ali', 'Sara', 'John', 'Zara', 'Mike']
count = 0

for user in users:
    if count == 3:
        print('Limit reached. Stop sending emails.')
        break
    print(f'Email sent to {user}')
    count += 1
```

### Q2. Stop monitoring when critical temperature is reached (for + break)

```python
temps = [22, 25, 30, 45, 28]
critical = 40

for t in temps:
    if t >= critical:
        print('Critical temperature reached! Stop monitoring.')
        break
    print(f'Temperature: {t}')
```

### Q3. Total price of items in a shopping cart (for)

```python
cart = [100, 250, 50, 300]
total = 0

for price in cart:
    total += price

print(f'Total Price: {total}')
```

### Q4. Skip customers with unpaid bills (for + continue)

```python
customers = [
    {'name': 'Ali', 'paid': True},
    {'name': 'Sara', 'paid': False},
    {'name': 'John', 'paid': True}
]

for customer in customers:
    if not customer['paid']:
        continue
    print('Access granted to:', customer['name'])
```

### Q5. Personalized emails (for)

### Q6. Extract domain names from email list (for)

```python
emails = ['ali@gmail.com', 'sara@yahoo.com', 'john@hotmail.com']
domains = []

for email in emails:
    domains.append(email.split('@')[1])

print('domains:', domains)
```

### Q7. Numbers divisible by 3 or 5 (list comprehension)

```python
numbers = list(range(1, 31))
result = [n for n in numbers if n % 3 == 0 or n % 5 == 0]

print('Original:', numbers)
print('Divisible by 3 or 5:', result)
```

### Q8. Multiplication table (nested for loop)

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for n in numbers:
    for m in multipliers:
        print(f'{n} x {m} = {n * m}')
```

### Q9. Ask for password until correct (while)

```python
correct_password = 'python123'

while True:
    password = input('Enter password: ')
    if password == correct_password:
        print('Access granted!')
        break
    print('Incorrect password. Try again.\n')
```

### Q10. Countdown from 5 to 1 (while)

```python
n = 5

while n >= 1:
    print(n)
    n -= 1

print("Time's up!")
```
